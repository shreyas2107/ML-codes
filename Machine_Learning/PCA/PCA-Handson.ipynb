```python
#split the data into dependents and independents column
x = df.drop('Target', axis = 1)
y = df['Target']
```

```python
x
```

```python
y
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)
x_scaled
```

```python
x_scaled = pd.DataFrame(x_scaled)
x_scaled.head()
```

```python
#Covariance matrix -> variance of one column with all other columns
#Find the co-variance matrix of x_scaled
cov = np.cov(x_scaled.T)
cov
```

```python
cov_df = pd.DataFrame(cov)
cov_df
```

```python
#get the eigenvalues and eigenvectors of the matrix
eig_val, eig_vec = np.linalg.eig(cov_df)
print(eig_val)
print(eig_vec)
```

```python
#How much variance each component captures
pd.DataFrame(eig_val)
```

```python
#eig_vec - new axis / direction (principal components)
pd.DataFrame(eig_vec)
```

```python
#total variance
tot = sum(eig_val)
print(tot)
```

```python
#wherever the percentage touches 90# -> means 90% variance
```

```python
var_exp = [(i/tot)*100 for i in sorted(eig_val,reverse = True)]
var_exp
```

```python
#Cumulative Explained Variance
cum_var_exp = np.cumsum(var_exp)
pd.DataFrame(cum_var_exp)
```

```python
plt.figure(figsize=(10,5))
plt.bar(range(len(var_exp)), var_exp, label="Individual Explained Variance", color="b")
plt.step(range(len(cum_var_exp)), cum_var_exp, label="Cummulative Explained Variance",color='r')
plt.xlabel("Principal Components")
plt.ylabel("Explained Variance Ration")
plt.show()
```

```python
x_train,x_test,y_train,y_test = train_test_split(x_pca,y,train_size = 0.8,random_state = 42)
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
model.fit(x_train,y_train)
```

```python
y_pred = model.predict(x_test)
y_pred
```

```python
from sklearn.metrics import *
accuracy_score(y_test, y_pred)
```

**An F1-score in a classification report is a single metric that combines precision and recall**

F1-Score: The happy middle ground (harmonic mean) between precision and recall, scored from 0 (worst) to 1 (best).

average='macro': Calculates the F1-score independently for each class and then takes the unweighted average [1]. Use this if you want to treat all classes equally, regardless of how frequent they are [1].

average='weighted': Calculates the F1-score for each class and takes the average weighted by the number of true instances (support) in each class [1]. Use this if your classes are imbalanced and you want to account for class size [1].

average='micro': Computes global precision and recall by counting the total true positives, false negatives, and false positives across all classes [1].

```python
f1_score(y_test,y_pred, average='macro')
```

```python
f1_score(y_test,y_pred, average='weighted')
```

```python
f1_score(y_test,y_pred, average='micro')
```
