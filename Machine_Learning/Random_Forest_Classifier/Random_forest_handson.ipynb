```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
y_test
```

```python
from imblearn.over_sampling import SMOTE
sm=SMOTE(random_state=42)
```

```python
#fit_resample - used to balance imbalanced datasets before training machine learning models. 
#It resolves class imbalances by oversampling minority classes, undersampling majority classes, or using a combination of both.
x_train_sm,y_train_sm=sm.fit_resample(x_train,y_train)
```

```python
x_train_sm
```

```python
y_train_sm
```

```python
print("Before SMOTE")
print(y_train)
print("After SMOTE")
print(y_train_sm)
```

```python
#Train the model
from sklearn.ensemble import RandomForestClassifier
model=RandomForestClassifier()
model.fit(x_train_sm,y_train_sm)
```

```python
#Test the Model
y_pred=model.predict(x_test)
```

```python
y_pred
```

```python
#Evaluate the Model
from sklearn.metrics import *
```

```python
accuracy_score(y_test,y_pred)
```

```python
precision_score(y_test,y_pred)
```

```python
recall_score(y_test,y_pred)
```

# Hyperparameters in Random Forest:

1. n_estimator:

It tells how many DT you are using to build the random forest model.

More trees=Better Accuracy.

```python
#use n_estimator=[50,100,150,200,250,300,350,400,450,500,550,600]
#Build the Random Forest model and tell which one is giving the highest accuracy

# 1. Define the list of n_estimators to test
n_estimators_list = [50, 100, 150, 200, 250, 300, 350, 400, 450, 500, 550, 600]

best_accuracy = 0
best_n = None

# 2. Loop through each value
for n in n_estimators_list:
    # Initialize the model with the current n_estimators
    # Note: using your balanced training data (x_train_sm, y_train_sm) from cell [42]
    rf_model = RandomForestClassifier(n_estimators=n, random_state=42, n_jobs=-1)
    
    # Fit the model
    rf_model.fit(x_train_sm, y_train_sm)
    
    # Predict on the test set
    y_pred_temp = rf_model.predict(x_test)
    
    # Calculate accuracy
    current_accuracy = accuracy_score(y_test, y_pred_temp)
    current_recall = recall_score(y_test, y_pred_temp)
    print(f"n_estimators = {n}: Accuracy = {current_accuracy:.4f} Recall Score = {current_recall:.4f}")
```
