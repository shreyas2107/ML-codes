```python
#Class distribution plot
sns.countplot(x="Class", data=df)
plt.xlabel("Class")
plt.ylabel("Count")
plt.title("Class distribution")
plt.show()
```

```python
#Transaction amount distribution - plot
sns.histplot(df["Amount"], bins=50, kde=True, color="blue")
plt.title("Transaction Amount Distribution")
plt.xlabel("Amount")
plt.ylabel("Frequency")
plt.show()
```

```python
# Decision tree model
#separating the target column
x = df.drop(columns = ['Class'])
y = df['Class']
```

```python
x
```

```python
y
```

```python
y.value_counts()
```

```python
#Splitting into train and test
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size = 0.20,random_state = 42,stratify=y)
x_train
```

```python
x_test
```

```python
y_train
```

```python
y_test
```

```python
#Building the Decision tree model
from sklearn.tree import DecisionTreeClassifier
dt_model = DecisionTreeClassifier(class_weight = 'balanced',random_state = 42)
dt_model.fit(x_train,y_train)
```

```python
#Make predictions
y_pred = dt_model.predict(x_test)
y_pred
```

```python
# Create a df to show actual vs predicted
df_comparision = pd.DataFrame({'Actual':y_test,'Predicted':y_pred})
df_comparision
```

```python
#Model evaluation
from sklearn.metrics import *
```

```python
accuracy_score(y_test,y_pred)
```

```python
precision_score(y_test,y_pred)
```

```python
recall_score(y_test,y_pred)
```

```python
#Confusion matrix
cm = confusion_matrix(y_test, y_pred)
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Reds",
    xticklabels=["Normal (0)", "Fraud (1)"],
    yticklabels=["Normal (0)", "Fraud (1)"],
)
plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

```python
#Classification report
print(classification_report(y_test, y_pred, target_names=["Normal", "Fraud"]))
```

```python
#Random Forest model
from sklearn.ensemble import RandomForestClassifier

# Initialize Random Forest Classifier
rf_model = RandomForestClassifier(n_estimators=50,criterion='entropy',max_depth=10,class_weight='balanced',random_state=42)
```

```python
# Fit model
rf_model.fit(x_train, y_train)
```

```python
# Make predictions
y_pred_rf = rf_model.predict(x_test)
```

```python
# Create a df to show actual vs predicted
df_comparison_rf = pd.DataFrame({'Actual':y_test,'Predicted':y_pred_rf})
df_comparison_rf
```

```python
accuracy_score(y_test,y_pred_rf)
```

```python
precision_score(y_test,y_pred_rf)
```

```python
recall_score(y_test,y_pred_rf)
```

```python
#Confusion matrix
cm_rf = confusion_matrix(y_test, y_pred_rf)
sns.heatmap(
    cm_rf,
    annot=True,
    fmt="d",
    cmap="Reds",
    xticklabels=["Normal (0)", "Fraud (1)"],
    yticklabels=["Normal (0)", "Fraud (1)"],
)
plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

```python
#Classification report
print(classification_report(y_test, y_pred_rf, target_names=["Normal", "Fraud"]))
```
