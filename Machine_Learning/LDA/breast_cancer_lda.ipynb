```python
# Feature separation
x = df.drop(columns = ['diagnosis'],errors = 'ignore')
y = df['diagnosis']
```

```python
# Label encoding for y - to change M (malignant) and B (benign) to 1 and 0
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
y_encoded = le.fit_transform(y)
print("Classes mapped:",le.classes_)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, accuracy_score

# 1. Train-test split
x_train, x_test, y_train, y_test = train_test_split(x, y_encoded, test_size=0.2, random_state=42)
```

```python
y_test
```

```python
# 2. Scale features
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
# 3. Fit Linear Discriminant Analysis
# Note: For binary classification (2 classes), n_components can at most be n_classes - 1 = 1
lda = LinearDiscriminantAnalysis(n_components=1)
x_train_lda = lda.fit_transform(x_train_scaled, y_train)
x_test_lda = lda.transform(x_test_scaled)
```

```python
# 4. Train a classifier on LDA features and evaluate
model = LogisticRegression()
model.fit(x_train_lda, y_train)
y_pred = model.predict(x_test_lda)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```

```python
from sklearn.ensemble import RandomForestClassifier

# Using LDA components with Random Forest
rf_lda = RandomForestClassifier(random_state=42)
rf_lda.fit(x_train_lda, y_train)

y_pred = rf_lda.predict(x_test_lda)
print("Random Forest + LDA Accuracy:", accuracy_score(y_test, y_pred))
```
