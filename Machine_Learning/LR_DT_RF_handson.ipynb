```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df['TotalCharges'].value_counts()
```

```python
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors = 'coerce')
```

```python
df['TotalCharges'].value_counts()
```

```python
df.isnull().sum()
```

```python
df['TotalCharges'] = df['TotalCharges'].fillna(df['tenure']*df['MonthlyCharges'])
```

```python
df.isnull().sum()
```

```python
df.drop(columns = ['customerID'], inplace = True)
```

```python
#Update categorical values to Numerical
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
```

```python
mapping = {}
for i in df.columns:
    if df[i].dtype == 'object':
        df[i] = le.fit_transform(df[i])
        mapping[i] = dict(zip(le.classes_,le.transform(le.classes_)))
print(mapping)
```

```python
df
```

```python
#Get the value counts for the churn
df['Churn'].value_counts()
```

```python
#Train test split
x = df.drop("Churn",axis = 1)
y = df['Churn']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.20,random_state = 42)
```

```python
x_train
```

```python
x_test
```

```python
y_train
```

```python
y_test
```

```python
#Logistic Regression
from sklearn.linear_model import LogisticRegression
model=LogisticRegression()
```

```python
from sklearn.feature_selection import RFE
#Apply rfe
rfe=RFE(model)
rfe.fit(x,y)
```

```python
#RFE will give the index of the columns that are important
rfe.support_  #support_ will give you a boolean list in which wherever you are having True means that column is imp
```

```python
imp_index_col=[i for i,data in enumerate(rfe.support_) if data==True]
```

```python
column_name=x.columns
imp_cols=[column_name[i] for i in imp_index_col]
imp_cols
```

```python
#test the model
model.fit(x_train,y_train)
#predict to test the model. Test the model
y_pred = model.predict(x_test)
y_pred
```

```python
#Evaluate the Model
#For a Binary Classification Problem:

 #Accuracy Score - (TP+TN)/(TP+TN+FP+FN) - TP: True Positive, TN: True Negative, FP: False Positive, FN: False negative
 #Precision Score - Out of all positives, how many were actually positive TP/(TP+FP)
 #Recall Score- Out of all positives, how many were actually predicted by the model TP/(TP+FN)
 #Confusion Matrix

from sklearn.metrics import *
```

```python
cm = confusion_matrix(y_test,y_pred) #output format [[TN,FP],[FN,TP]]
```

```python
cm
```

```python
accuracy_score(y_test,y_pred)
```

```python
precision_score(y_test,y_pred)
```

```python
recall_score(y_test,y_pred)
```

```python
#test the model
y_pred = dt_model.predict(x_test)
y_pred
```

```python
# Hyperparameter tuning
tree_model = DecisionTreeClassifier()
```

```python
#best hyper-parameter
#Grid Search CV
```

```python
param_grid = {
    "criterion" : ["gini","entropy"],
    "max_depth" : [3,4,6,8,9,10],
    "min_samples_split" : [2,5,10],
    "min_samples_leaf" : [1,2,4]
}
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
grid_search = GridSearchCV(
    estimator = tree_model,
    param_grid = param_grid,
    cv = 5, #cross validation - data will be split into 5 parts (cv = 4 or 5 is sufficient)
    scoring = "accuracy"
)
grid_search.fit(x_train,y_train)
```

```python
#best hyperparameter
print("Best Hyperparameter:",grid_search.best_params_)
```

```python
#best score
print("Best Score:",grid_search.best_score_)
```

```python
#Train the model
from sklearn.ensemble import RandomForestClassifier
rf_model=RandomForestClassifier()
param_grid = {
    "n_estimators" : [100,150,200],
    "criterion" : ["gini","entropy"],
    "max_depth" : [3,4,6,8,9,10],
    "min_samples_split" : [2,5,10],
    "min_samples_leaf" : [1,2,4],
    "random_state" : [0,42]
}
grid_search = GridSearchCV(
    estimator = rf_model,
    param_grid = param_grid,
    cv = 5, #cross validation - data will be split into 5 parts (cv = 4 or 5 is sufficient)
    scoring = "accuracy"
)

grid_search.fit(x_train_sm,y_train_sm)
```
