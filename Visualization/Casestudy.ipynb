```python
#creating the countplot to get the number of projects
sns.countplot(x='numberOfProjects',data = df,color = 'yellow')
plt.show()
```

```python
#department vs attrition
sns.countplot(data= df,x='dept',hue='left')
plt.show()
```

```python
#Correlation heatmap
#heatmap
cor_mat = df.corr(numeric_only = True)
print(cor_mat)
```

```python
#to get the correlation data in the heatmap
sns.heatmap(cor_mat,annot=True, cmap="coolwarm")
plt.show()
```

```python
#salary vs satisfaction
sns.boxplot(data= df,x='salary',y='satisfactoryLevel',hue = 'left')
plt.show()
```
