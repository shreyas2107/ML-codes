```python
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
dict1 = {'days':['M','Tu','W','Th','F','Sa','Su'],
 'sales':[200,250,275,280,295,325,350],
 'visitors':[50,45,60,70,85,90,120],
 'gender':['F','M','F','M','F','M','F']
}
data = pd.DataFrame(dict1)
data
```

```python
#Lineplot
sns.lineplot(x='days',y='sales',data=data,marker = '*',markersize = 10)
plt.show()
```

```python
#Hue differentiates the data based on the column data present
sns.lineplot(x='days',y='sales',data=data,hue = 'gender',marker = '*',markersize = 10)
plt.show()
```

```python
#Barplot
sns.barplot(x='days',y='sales',data=data,color = 'green')
plt.title("Sales per day")
plt.show()
```

```python
sns.barplot(x='days',y='sales',hue = 'gender',data=data,color = 'green')
plt.title("Sales per day")
plt.show()
```

```python
sns.barplot(x='days',y='sales',palette=['pink','yellow','red','cyan','purple'],data=data,edgecolor='grey')
plt.title('BAR PLOT USING SEABORN')
plt.show()
```

```python
#Histogram
sns.histplot(data['visitors'],color = 'blue')
plt.show()
```

```python
#scatterplot
sns.scatterplot(x='days',y='sales',hue = 'gender',data=data,color = 'green')
plt.title("Sales per day")
plt.show()
```

```python
#boxplot
sns.boxplot(data['sales'])
plt.show()
```

```python
#Violinplot
sns.violinplot(data['visitors'])
plt.show()
```

```python
#Countplot
sns.countplot(x='gender',data = data,color = 'yellow')
plt.show()
```

```python
#heatmap
cor_mat = data.corr(numeric_only = True)
print(cor_mat)
```

```python
sns.heatmap(cor_mat)
plt.show()
```

```python
#to get the correlation data in the heatmap
sns.heatmap(cor_mat,annot=True, cmap="coolwarm")
```

```python
#Pair plot
sns.pairplot(data)
plt.show()
```
