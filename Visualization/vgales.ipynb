```python
#Remove null values
df['Year'] = df['Year'].fillna(0)
df.isnull().sum()
```

```python
df['Publisher'] = df['Publisher'].fillna("Unknown")
df.isnull().sum()
```

```python
df.info()
```

```python
df['Year'] = df['Year'].astype('int')
```

```python
df.dtypes
```

```python
df['Year'] = pd.to_numeric(df['Year'], errors = 'coerce')
df.dtypes
```

```python
df['Year'].value_counts()
```

```python
#2020 has only one value - can drop the row
index_to_drop = df[df['Year'] ==2020].index
index_to_drop
```

```python
df.drop(index_to_drop,inplace = True)
```

```python
#Top 10 games with Global sales
top_10_games = df.nlargest(10, 'Global_Sales')

# Display the result
top_10_games
```

```python
df[['Name', 'Global_Sales']].sort_values(by='Global_Sales', ascending=False).head(10)
```

```python
plt.figure(figsize = (8,6))
sns.barplot(x = 'Global_Sales', y = 'Name' , data = top_10_games)
plt.title("Top 10 Best Selling Games Globally")
plt.show()
```

```python
#Which genre is most popular globally
genre_sales = df.groupby('Genre')['Global_Sales'].sum().sort_values(ascending=False)
genre_sales
```

```python
plt.figure(figsize = (8,6))
sns.barplot(x = genre_sales.values, y = genre_sales.index)
plt.title("Most Popular Genres Globally")
plt.show()
```

```python
#Which Publisher has the highest global sales
top_publisher_sales = df.groupby('Publisher')['Global_Sales'].sum().sort_values(ascending=False).head(10)
top_publisher_sales
```

```python
plt.figure(figsize = (8,6))
sns.barplot(x = top_publisher_sales.values, y = top_publisher_sales.index)
plt.title("Publishers with Highest Global Sales")
plt.show()
```

```python
#Which region contributes more to the global sales
# Summing the sales for each region
regional_sales = df[['NA_Sales', 'EU_Sales', 'JP_Sales', 'Other_Sales']].sum()

# Sorting to see which is highest
regional_sales = regional_sales.sort_values(ascending=False)
regional_sales
```

```python
import plotly.express as px
fig = px.pie(values = regional_sales.values,names = regional_sales.index, title = "Region-wise Contribution to Global sales")
fig.show()
```

```python
#Chart that shows the trend of global over time period
# Summing the global sales for each year
df1 = df[df['Year'] != 0]
yearly_global_sales = df1.groupby('Year')['Global_Sales'].sum().reset_index()
yearly_global_sales
```

```python
fig = px.line(yearly_global_sales, x = 'Year', y = 'Global_Sales', title = "Global Sales Trend per year")
fig.show()
```
