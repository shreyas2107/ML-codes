```python
import matplotlib.pyplot as plt
```

**Line plot**

```python
sales = [500,340,450,350,650,700,540]
days = ['M','Tue','W','Th','F','Sat','Sun']
plt.plot(days,sales)
plt.show()
```

```python
plt.xlabel("Days of the week")
plt.ylabel("Ice-cream sales per day")
plt.title("Ice-cream sales for the week")
plt.grid() # to add grid for better readability
plt.plot(days,sales)
plt.show()
```

```python
# to add marker, color, linestyle, linewidth
plt.plot(days,sales,marker = 'o',color = 'red',linestyle = '-.',linewidth= 4)
plt.grid()
plt.show()
```

```python
# mobile sales per year
mob1 = [250,300,325,375,450,500,540]
mob2 = [200, 350, 270, 450, 500, 450, 370]
year = [2020,2021,2022,2023,2024,2025,2026]
plt.plot(year,mob1,marker = 'd',color = 'red',linestyle = 'dotted',linewidth= 2,label= 'Mobile 1') # label is legend
plt.plot(year,mob2,marker = 'd',color = 'violet',linestyle = 'dotted',linewidth= 2,label= 'Mobile 2')
plt.xlabel("Sales year")
plt.ylabel("Mobile sales")
plt.title("Mobile sales per year")
#can move legend as we need
plt.legend(loc='upper left')
plt.grid()
plt.show()
```

**Bar Plot**

```python
product = ['Mango','Choco','Kiwi']
sales = [500,700,350]
plt.bar(product,sales,color = ['yellow','brown','green'])
plt.xlabel("Ice-cream flavor")
plt.ylabel("Ice-cream sales")
plt.title("Ice-cream sales per flavor")
plt.grid(False)
plt.show()
```

```python
# Clinical trial
patient_groups = ['Placebo','Single-blind','Double-blind','Test group']
patient_count = [150,200,100,125]
plt.bar(patient_groups,patient_count,color = 'blue')
plt.xlabel("Subject groups")
plt.ylabel("Subjects enrolled")
plt.title("Subjects enrolled per Subject group")
plt.grid(False)
plt.show()
```

**Pie plots**

```python
fruits = ['Mango','Apple','Orange']
votes = [50,45,48]
plt.pie(votes,labels= fruits, autopct = "%0.2f%%")
plt.show()
```

```python
# website traffic -- number of users visiting the website

sources = ["friends", "social_media", "wikipedia", "newspaper", "billboard", "workplace"]
visitors = [20, 70, 20, 5, 15, 20]

plt.pie(visitors, labels = sources, autopct = "%0.2f%%", explode = [0, 0, 0, 0.2, 0, 0])
plt.show()
```

```python
#area plot
sales = [500,340,450,350,650,700,540]
days = ['M','Tue','W','Th','F','Sat','Sun']
plt.plot(days,sales,color = 'red')
plt.fill_between(days,sales,color = 'yellow')
plt.xlabel("Days of the week")
plt.ylabel("Ice-cream sales per day")
plt.title("Ice-cream sales for the week")
plt.grid()
plt.show()
```

```python
#Subplots
plt.subplots(1,2,figsize=(10,3)) # 10 is width , 5 is height


# 1st subplot
plt.subplot(1,2,1)
days = ['mon','tue','wed','thur','fri']
sales = [460,290,380,480,525]
plt.plot(days,sales,color='magenta',marker='*',markerfacecolor='seagreen')
plt.title('1st subplot')


# 2nd subplot
plt.subplot(1,2,2)
days = ['mon','tue','wed','thur','fri']
sales = [460,290,380,480,525]
plt.bar(days,sales,color='pink',edgecolor='cyan')
plt.title('2nd subplot')
plt.show()
```
