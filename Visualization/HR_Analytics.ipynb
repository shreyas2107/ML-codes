```python
#Find the avg salary
print(df["MonthlyIncome"].mean())
#Find the avg salary department wise
df.groupby("Department")["MonthlyIncome"].mean()
```

```python
#Find the count of male and female in each department
df.groupby("Department")["Gender"].value_counts().unstack()
```

```python
#Visualizations
#Find the avg salary dept wise using a bar chart
avg_sal_dept = df.groupby("Department")["MonthlyIncome"].mean()
sns.barplot(x = "Department", y = "MonthlyIncome",data = df,errorbar = None, hue = "Department")
plt.title('Average Monthly Income by Department')
plt.show()
```

```python
#Find the average years at company,job role wise using horizontal bar chart
avg_years_role = df.groupby('JobRole')['YearsAtCompany'].mean()
avg_years_role
```

```python
plt.barh(avg_years_role.index,avg_years_role)
plt.title("Job role-wise average years at company")
plt.xlabel("YearsAtCompany")
plt.ylabel("JobRole")
plt.show()
```

```python
df['JobSatisfaction'].unique()
```

```python
# Find the Gender-wise job satisfaction level distribution
plt.figure(figsize = (4,3))
sns.countplot(x="JobSatisfaction",data=df,hue="Gender")
plt.show()
```

```python
#Find the attrition percentage
# Calculate the count of each Attrition category
attrition_counts = df['Attrition'].value_counts()

# Plotting the pie chart
plt.figure(figsize=(6, 6))
plt.pie(attrition_counts, 
        labels=attrition_counts.index, 
        autopct='%0.2f%%', 
        explode=[0,0.1],
        startangle=180,
        wedgeprops={'edgecolor': 'black', 'linewidth': 3}) 

# Adding title
plt.title('Attrition Percentage Distribution', fontsize=14, fontweight='bold')

# Display the plot
plt.show()
```

```python
#Find the avg salary by age using line plot
sns.lineplot(data=df,x='Age',y='MonthlyIncome',marker='d',errorbar=None)
plt.title("Average Salary by Age")
plt.show()
```

```python
#Which department people are more likely to leave the company and which department people are less likely
pd.crosstab(df['Department'], df['Attrition'], normalize='index') * 100
```

```python
sns.countplot(x = 'Department',hue='Attrition',data = df)
plt.title("Attrition rate per department")
plt.show()
```

```python
#Which job role is more likely to leave the company and which job role less
pd.crosstab(df['JobRole'], df['Attrition'], normalize='index') * 100
```

```python
plt.figure(figsize = (9,4))
sns.countplot(x = 'JobRole',hue='Attrition',data = df)
plt.xticks(rotation=45)
plt.title("Attrition rate per Job Role")
plt.show()
```

```python
#Find the distribution of the salary
sns.histplot(x="MonthlyIncome",data=df,edgecolor='Black',lw=2)
plt.show()
```

```python
#Find the Attrition according to Years at the company
sns.boxplot(y='YearsAtCompany',hue= 'Attrition',data = df)
plt.title("Attrition according to years at company")
plt.show()
```

```python
sns.histplot(x='YearsAtCompany',hue= 'Attrition',data = df)
plt.title("Attrition according to years at company")
plt.show()
```

```python
#Years since last promotion vs Attrition
sns.countplot(x='YearsSinceLastPromotion',hue='Attrition',data=df)
plt.title("Years since Last Promotion vs Attrition")
plt.show()
```

```python
#Create a plot to show Salary hike vs Performance rating
sns.boxplot(x='PerformanceRating',y='PercentSalaryHike',data = df)
plt.title("Salary Hike vs Performance Rating")
plt.show()
```

```python
# Training time vs Performance Rating
sns.boxplot(x='PerformanceRating',y='TrainingTimesLastYear',data = df)
plt.title("Training Time vs Performance Rating")
plt.show()
```
