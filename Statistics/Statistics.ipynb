```python
# Q2. Two students have marks 
A = [78, 80, 82, 79, 81]
B = [50, 100, 60, 90, 80]


# a. calculate mean
a_mean = stats.mean(A)
print(a_mean)
b_mean = stats.mean(B)
print(b_mean)
# b. calculate var and std dev
a_var = stats.variance(A)
print(a_var)
b_var = stats.variance(B)
print(b_var)
a_std = np.std(A)
print(a_std)
b_std = np.std(B)
print(b_std)
# c. which student is getting more marks and which is more consistent?
```
