```python
#3.Find the range of population of all countries in 2017
pop_range = df['Population in thousands (2017)'].max() - df['Population in thousands (2017)'].min()
pop_range
```

```python
#4. Find the inter quartile range of the population of all countries in 2017
pop_q1 = np.percentile(df['Population in thousands (2017)'],25)
pop_q3 = np.percentile(df['Population in thousands (2017)'],75)
iqr = pop_q3 - pop_q1
iqr
```
