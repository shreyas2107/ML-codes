```python
#IQR
a = [25, 27, 29, 30, 32, 34, 35]
b = [25, 27, 29, 30, 32, 34, 100]
```

```python
q1_a = np.percentile(a,25)
q3_a = np.percentile(a,75)
iqr_a = q3_a - q1_a
iqr_a
```

```python
q1_b = np.percentile(b,25)
q3_b = np.percentile(b,75)
iqr_b = q3_b - q1_b
iqr_b
```

```python
c = [10, 24, 25, 26, 37, 34, 32, 30, 28, 34, 33, 140]
```

```python
quar1 = np.percentile(c,25)
quar3 = np.percentile(c,75)
iqr = quar3 - quar1
iqr
```

```python
#upper limit
ul = quar3 + 1.5*iqr
ul
```

```python
#lower limit
ll = quar1 - 1.5*iqr
ll
```

```python
clean_data = []
for val in c:
    if val >= ul or val <= ll:
        print("Value is an outlier")
    else:
        clean_data.append(val)
clean_data
```
