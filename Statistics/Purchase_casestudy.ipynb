**1) It was observed that the average purchase made by the Men of the age 18-25 was 10000. Is it correct/still same?**

```python
data = df[(df["Gender"]=="M") & (df["Age"]=="18-25")]
data
```

```python
# Sampling

sample= data.sample(3600, random_state=42)
sample
```

```python
# check the sample mean
print(sample["Purchase"].mean())
```

**Null Hypothesis(H0)**

**Sample mean equal to 10000**

**Alernate Hypothesis(H1)**

**Sample mean not equal to 10000**

```python
assumed_mean = 10000
```

```python
# We apply one-sample t-test here

from scipy.stats import ttest_1samp

t_stats, p_val = ttest_1samp(sample['Purchase'], assumed_mean)
print("T-stats:",t_stats)
print(f"P-Value: {p_val: .30f}")
```

```python
if p_val < 0.5:
    print("Reject the null hypothesis -> Sample mean not equal to 10000")
else:
    print("Accept the null hypothesis -> Sample mean equal to 10000")
```

**It was observed that the percentage of women that spend more than 10000. Is it still the same ?**

```python
data= df[(df["Purchase"]>10000)]
data.shape
```

```python
total= 91057
```

```python
# Number of women in data
count= data["Gender"].value_counts()
count
```

```python
count_women= 18685 # Number of women
p0 = 0.35
```

**H0 -> Proportion is 35%
H1 -> Proportion is not 35%**

```python
# Z test for porportion
#Manual check
print((count_women/total)*100,"%")
```

```python
# Z test
from statsmodels.stats.proportion import proportions_ztest

z_stats,p_val = proportions_ztest(count= count_women, nobs= total, value = p0)
print("Z-stats:",z_stats)
print(f"P-Value: {p_val: .30f}")
```

```python
if p_val < 0.5:
    print("Reject the null hypothesis -> Proportion is not 35%")
else:
    print("Accept the null hypothesis -> Proportion is 35%")
```
